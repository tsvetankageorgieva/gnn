```python
#vЗареждане на набора от данни
from torch_geometric.datasets import Planetoid

dataset = Planetoid(root="data/Cora", name="Cora")
data = dataset[0]
```

```python
# Конструиране на GAT модел
import torch
import torch.nn.functional as F
from torch_geometric.nn import GATConv
from torch_geometric.nn import GCNConv

class GAT(torch.nn.Module):
    def __init__(self):
        super().__init__()

        self.conv1 = GATConv(
            dataset.num_features,
            8,
            heads=8,
            dropout=0.6)

        self.conv2 = GATConv(
            8 * 8,
            dataset.num_classes,
            heads=1,
            concat=False,
            dropout=0.6)

    def forward(self, data):
        x, edge_index = data.x, data.edge_index

        x = self.conv1(x, edge_index)
        x = F.elu(x)
        x = F.dropout(x, p=0.6, training=self.training)

        x = self.conv2(x, edge_index)

        return x
```

```python
# Създаване на модела
model = GAT()
```

```python
# Дефиниране на оптимизатор
optimizer = torch.optim.Adam(
    model.parameters(),
    lr=0.005,
    weight_decay=5e-4)
```

```python
# Функция на загубите
criterion = torch.nn.CrossEntropyLoss()
```

```python
# Обучение на модела
num_epochs = 200

train_loss = []
train_acc = []
test_acc = []

for epoch in range(num_epochs):
    model.train()
    optimizer.zero_grad()

    out = model(data)

    loss = criterion(out[data.train_mask],
        data.y[data.train_mask] )

    loss.backward()
    optimizer.step()
    train_loss.append(loss.item())
    model.eval()

    with torch.no_grad():
        pred = out.argmax(dim=1)
        train_correct = (pred[data.train_mask] == 
            data.y[data.train_mask]).sum()

        train_accuracy = (train_correct /
            data.train_mask.sum()).item()

        train_acc.append(train_accuracy)

        test_correct = (pred[data.test_mask] ==
            data.y[data.test_mask]).sum()

        test_accuracy = (test_correct /
            data.test_mask.sum()).item()

        test_acc.append(test_accuracy)
```

```python
# Изчисляване на Accuracy
model.eval()
with torch.no_grad():
    out = model(data)
    pred = out.argmax(dim=1)

    correct = (pred[data.test_mask] ==
        data.y[data.test_mask]).sum()
    accuracy = (correct / data.test_mask.sum()).item()

print(f"Accuracy: {accuracy:.4f}")
```

```python
# Изчисляване на F1-score
from sklearn.metrics import f1_score

model.eval()

with torch.no_grad():
    out = model(data)
    pred = out.argmax(dim=1)

f1 = f1_score(
    data.y[data.test_mask].cpu(),
    pred[data.test_mask].cpu(),
    average="macro")

print(f"F1-score: {f1:.4f}")
```
